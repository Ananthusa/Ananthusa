```python
import requests
import json
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Example: Blacklisted Bitcoin Addresses (Dark Web, Fraud, Ransomware)
BLACKLISTED_ADDRESSES = {
    "1HQ3Go3ggs8pFnXuHVHRytPCq5fGG8Hbhx",  # Example dark web wallet
    "1Ez69SnzzmePmZX3WpEzMKTrcBF2gpNQ55",  # Silk Road seizure
}

# Sample dataset (transaction history) - Normally, use real blockchain data
data = {
    "tx_id": ["tx1", "tx2", "tx3", "tx4"],
    "amount": [0.5, 50.0, 1000.0, 200.0],  # BTC amount
    "num_inputs": [1, 5, 10, 2],  # Number of inputs
    "num_outputs": [1, 3, 8, 2],  # Number of outputs
    "blacklisted_involved": [0, 1, 1, 0],  # If a blacklisted address is involved (1 = Yes)
    "suspicious": [0, 1, 1, 0],  # Label (1 = Fraud, 0 = Safe)
}

df = pd.DataFrame(data)

# Features & Labels
X = df[["amount", "num_inputs", "num_outputs", "blacklisted_involved"]]
y = df["suspicious"]

# Train ML Model
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Function to analyze a Bitcoin address
def analyze_bitcoin_address(address):
    url = f"https://blockchain.info/rawaddr/{address}"
    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()
        transactions = data.get("txs", [])

        print(f"Checking address: {address}")
        for tx in transactions:
            inputs = [inp["prev_out"]["addr"] for inp in tx["inputs"] if "prev_out" in inp]
            outputs = [out["addr"] for out in tx["out"] if "addr" in out]

            # Features extraction
            amount = sum(out["value"] for out in tx["out"]) / 1e8  # Convert Satoshi to BTC
            num_inputs = len(inputs)
            num_outputs = len(outputs)
            blacklisted_involved = int(any(addr in BLACKLISTED_ADDRESSES for addr in inputs + outputs))

            # Prepare data for ML model
            features = scaler.transform([[amount, num_inputs, num_outputs, blacklisted_involved]])
            prediction = model.predict(features)[0]

            if prediction == 1:
                print(f"[⚠️ ALERT] Suspicious transaction detected at {tx['hash']} | Amount: {amount} BTC")
            else:
                print(f"[✔️ SAFE] Transaction {tx['hash']} is normal.")
    else:
        print("Error fetching data. Check API limits.")

# Example: Check multiple Bitcoin addresses
addresses_to_check = [
    "1BoatSLRHtKNngkdXEeobR76b53LETtpyT",  # Random BTC address
    "1Ez69SnzzmePmZX3WpEzMKTrcBF2gpNQ55",  # Blacklisted
]

for addr in addresses_to_check:
    analyze_bitcoin_address(addr)
```
